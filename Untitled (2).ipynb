# ⚙️ Feature Engineering
- Extracted Year, Month, and Day from the match date for time-based analysis.
- Derived Weekday and Day Name to analyze matches played on different days of the week.
- Converted the Date column into datetime format for efficient date-based analysis and visualization.


```python
df
```

```python
print(df['Date'].dtype)
```

```python
df['year'] = df['Date'].dt.year
df['month'] = df['Date'].dt.month
df['day'] = df['Date'].dt.day
df['weekday'] = df['Date'].dt.weekday
df['day_name'] = df['Date'].dt.day_name()

df[['Date','year','month','day','weekday','day_name']].head(11)
```

```python
# Weekend or Weekday
df['is_weekend'] = df['day_name'].isin(['Saturday', 'Sunday']).astype(int)

# Match category
def match_type(x):
    if x in ['Saturday', 'Sunday']:
        return 'Weekend Match'
    else:
        return 'Weekday Match'

df['match_type'] = df['day_name'].apply(match_type)

# Display
df[['Date','year','month','day','weekday','day_name','is_weekend','match_type']].head(10)
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Convert scores to numeric (extract only runs)
df['First Innings Score'] = pd.to_numeric(
    df['First Innings Score'].astype(str).str.extract(r'(\d+)')[0],
    errors='coerce'
)

df['Second Innings Score'] = pd.to_numeric(
    df['Second Innings Score'].astype(str).str.extract(r'(\d+)')[0],
    errors='coerce'
)

# Create correlation heatmap
plt.figure(figsize=(6,5))

sns.heatmap(
    df[['First Innings Score','Second Innings Score']].corr(),
    annot=True,
    cmap='coolwarm'
)

plt.title("Correlation Matrix")
plt.show()
```

# ⏳ Time Series Analysis
- Match trends over months
- Peak match month
- Year-wise match trends

```python
import pandas as pd

# Read IPL dataset
df = pd.read_csv("../data/IPL23dataset12.csv")

# Check columns
print(df.columns)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
df['Date'] = (
    df['Date']
    .str.replace(r'(\d+)(st|nd|rd|th)', r'\1', regex=True)
    .str.replace(r',.*', '', regex=True)
)

df['Date'] = pd.to_datetime(df['Date'])
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Convert Date column to datetime (if not already)
df['Date'] = pd.to_datetime(df['Date'])

# Set Date as index
df.set_index('Date', inplace=True)

# Daily match trend
daily_trend = df.resample('D').size()

plt.figure(figsize=(8,5))

daily_trend.plot()

plt.title("Daily Match Trend")
plt.xlabel("Date")
plt.ylabel("Number of Matches")

plt.show()
```

# 📍 Venue-Based Analysis
- Top Match Venues
- Venue-wise Winners

```python
import seaborn as sns
import matplotlib.pyplot as plt

venue = df['Venue'].value_counts().reset_index()
venue.columns = ['Venue', 'Matches']

plt.figure(figsize=(12,6))

sns.scatterplot(
    data=venue,
    x='Venue',
    y='Matches',
    s=150
)

plt.title("Venue-wise Match Distribution")
plt.xlabel("Venue")
plt.ylabel("Number of Matches")

plt.xticks(rotation=90)

plt.show()
```

```python
df['Winning Team'].value_counts().head(5).plot(
    kind='pie',
    autopct='%1.1f%%',
    figsize=(7,7)
)

plt.title("Top 5 Winning Teams")
plt.ylabel("")
plt.show()
```
